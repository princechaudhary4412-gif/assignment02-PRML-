```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('advertising.csv')

print(df.head())
print(df.shape)
print(df.columns)
```

```python
X = df[['TV', 'Radio', 'Newspaper']].values
y = df['Sales'].values

print("Input shape:", X.shape)
print("Output shape:", y.shape)
```

```python
np.random.seed(42)

indices = np.random.permutation(len(X))
split = int(0.8 * len(X))

train_indices = indices[:split]
test_indices = indices[split:]

X_train = X[train_indices]
X_test = X[test_indices]

y_train = y[train_indices]
y_test = y[test_indices]

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# Add a column of 1s for the intercept
X_train_new = np.c_[np.ones(len(X_train)), X_train]

# Calculate regression coefficients
coefficients = np.linalg.pinv(X_train_new) @ y_train

print("Regression coefficients:")
print(coefficients)
```

```python
# Add a column of 1s to the testing data
X_test_new = np.c_[np.ones(len(X_test)), X_test]

# Predict sales
y_pred = X_test_new @ coefficients

print("Actual sales:")
print(y_test[:5])

print("Predicted sales:")
print(y_pred[:5])
```

```python
# R2 Score
ss_res = np.sum((y_test - y_pred) ** 2)
ss_total = np.sum((y_test - np.mean(y_test)) ** 2)

r2 = 1 - (ss_res / ss_total)

# Mean Absolute Error
mae = np.mean(np.abs(y_test - y_pred))

# Mean Squared Error
mse = np.mean((y_test - y_pred) ** 2)

print("Advertising Dataset Results")
print("R2 Score:", r2)
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
```

```python
df2 = pd.read_csv('Housing.csv')

print(df2.head())
print(df2.shape)
print(df2.columns)
print(df2.dtypes)
```

```python
X2 = df2.drop('price', axis=1)
y2 = df2['price'].values

print("Input data:")
print(X2.head())

print("Target data:")
print(y2[:5])
```

```python
X2 = pd.get_dummies(X2, drop_first=True, dtype=int)

X2 = X2.values

print("Shape after conversion:", X2.shape)
print(X2[:5])
```

```python
np.random.seed(42)

indices2 = np.random.permutation(len(X2))
split2 = int(0.8 * len(X2))

train_indices2 = indices2[:split2]
test_indices2 = indices2[split2:]

X2_train = X2[train_indices2]
X2_test = X2[test_indices2]

y2_train = y2[train_indices2]
y2_test = y2[test_indices2]

print("Training rows:", len(X2_train))
print("Testing rows:", len(X2_test))
```

```python
# Add a column of 1s for the intercept
X2_train_new = np.c_[np.ones(len(X2_train)), X2_train]

# Calculate regression coefficients
coefficients2 = np.linalg.pinv(X2_train_new) @ y2_train

print("Housing regression coefficients:")
print(coefficients2)
```

```python
X2_test_new = np.c_[np.ones(len(X2_test)), X2_test]

y2_pred = X2_test_new @ coefficients2

print("Actual prices:")
print(y2_test[:5])

print("Predicted prices:")
print(y2_pred[:5])
```

```python
# R2 Score
ss_res2 = np.sum((y2_test - y2_pred) ** 2)
ss_total2 = np.sum((y2_test - np.mean(y2_test)) ** 2)

r2_2 = 1 - (ss_res2 / ss_total2)

# Mean Absolute Error
mae2 = np.mean(np.abs(y2_test - y2_pred))

# Mean Squared Error
mse2 = np.mean((y2_test - y2_pred) ** 2)

print("Housing Dataset Results")
print("R2 Score:", r2_2)
print("Mean Absolute Error:", mae2)
print("Mean Squared Error:", mse2)
```

```python
results = pd.DataFrame({
    'Dataset': ['Advertising', 'Housing'],
    'R2 Score': [r2, r2_2],
    'MAE': [mae, mae2],
    'MSE': [mse, mse2]
})

print(results.to_string(index=False))
```

```python
import pandas as pd
d1=pd.read_csv("validationData.csv")
d2=pd.read_csv("trainingData.csv")
df= pd.concat([d1, d2], ignore_index=True)

print(df.head())
```

```python
import numpy as np
import pandas as pd


data = df

targets = ["LATITUDE", "LONGITUDE", "FLOOR"]

def correlation(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    x_mean = np.mean(x)
    y_mean = np.mean(y)

    numerator = np.sum((x - x_mean) * (y - y_mean))

    denominator = np.sqrt(
        np.sum((x - x_mean) ** 2) *
        np.sum((y - y_mean) ** 2)
    )

    if denominator == 0:
        return 0

    return numerator / denominator

target = data[["LATITUDE", "LONGITUDE", "FLOOR"]]

finaldata= data.drop(columns=["LATITUDE", "LONGITUDE", "FLOOR"])

c=finaldata.columns
count =0
pairs=[]
for i in range(len(c)):
    for j in range(i + 1, len(c)):

        r = correlation(data[c[i]], data[c[j]])

        if abs(r) > 0.75:
            pairs.append((c[i], c[j]))
            count++

print(pairs)
print(count)

```

```python
removed = []

for col1, col2 in pairs:

    score1 = 0
    score2 = 0

    for target_col in target.columns:
        score1 += abs(correlation(data[col1], target[target_col]))
        score2 += abs(correlation(data[col2], target[target_col]))

    if score1 < score2:
        removed.append(col1)
        print("Removed:", col1)
    else:
        removed.append(col2)
        print("Removed:", col2)

data=data.drop(columns=removed)
remain_data=data.copy()
```

```python
import pandas as pd
import math

df1 = pd.read_csv('advertising.csv')

target = 'Sales'
features = [col for col in df1.columns if col != target]


def correlation(x, y):
    x_mean = sum(x) / len(x)
    y_mean = sum(y) / len(y)

    numerator = 0
    x_square = 0
    y_square = 0

    for i in range(len(x)):
        numerator += (x[i] - x_mean) * (y[i] - y_mean)
        x_square += (x[i] - x_mean) ** 2
        y_square += (y[i] - y_mean) ** 2

    return numerator / math.sqrt(x_square * y_square)



drop_columns = set()


threshold = 0.91


for i in range(len(features)):
    for j in range(i + 1, len(features)):

        col1 = features[i]
        col2 = features[j]

        r = correlation(df1[col1], df1[col2])

        print(col1, "vs", col2, "=", r)


        if abs(r) >= threshold:


            r1_target = correlation(df1[col1], df1[target])
            r2_target = correlation(df1[col2], df1[target])

            print("  HIGH CORRELATION")
            print(col1, "with", target, "=", r1_target)
            print(col2, "with", target, "=", r2_target)


            if abs(r1_target) < abs(r2_target):
                drop_columns.add(col1)
                print("  -->", col1, "can be dropped")

            else:
                drop_columns.add(col2)
                print("  -->", col2, "can be dropped")



drop_columns = list(drop_columns)

print("\nColumns to be dropped:")
print(drop_columns)



final_df = df1.drop(columns=drop_columns)

print("\nFinal dataset:")
print(final_df.head())

print("\nRemaining columns:")
print(final_df.columns)



final_df.to_csv('advertising_final.csv', index=False)
```

```python
import pandas as pd
import math

df1 = pd.read_csv('advertising.csv')
df2 = pd.read_csv('Housing.csv')

# Target
target2 = df2['price']

# Remove target from features
df2f = df2.drop('price', axis=1)

# Yes/No columns
yes_no_columns = [
    'mainroad',
    'guestroom',
    'basement',
    'hotwaterheating',
    'airconditioning',
    'prefarea'
]

# Convert yes/no to 1/0
for col in yes_no_columns:
    df2f[col] = df2f[col].map({'yes': 1, 'no': 0})


# If furnishingstatus is present, convert it too
df2f['furnishingstatus'] = df2f['furnishingstatus'].map({
    'unfurnished': 0,
    'semi-furnished': 1,
    'furnished': 2
})


# All remaining columns are features
features2 = list(df2f.columns)


def correlation(x, y):

    x_mean = sum(x) / len(x)
    y_mean = sum(y) / len(y)

    numerator = 0
    x_square = 0
    y_square = 0

    for i in range(len(x)):
        numerator += (x[i] - x_mean) * (y[i] - y_mean)
        x_square += (x[i] - x_mean) ** 2
        y_square += (y[i] - y_mean) ** 2

    return numerator / math.sqrt(x_square * y_square)


drop_columns = set()

threshold = 0.80


# Compare every feature with every other feature
for i in range(len(features2)):

    for j in range(i + 1, len(features2)):

        col1 = features2[i]
        col2 = features2[j]

        r = correlation(df2f[col1], df2f[col2])

        print(col1, "vs", col2, "=", r)

        if abs(r) >= threshold:

            # Compare both features with TARGET
            r1_target = correlation(df2f[col1], target2)
            r2_target = correlation(df2f[col2], target2)

            print("  HIGH CORRELATION")

            print(col1, "with price =", r1_target)
            print(col2, "with price =", r2_target)

            # Drop feature having lower target correlation
            if abs(r1_target) < abs(r2_target):

                drop_columns.add(col1)

                print("  -->", col1, "can be dropped")

            else:

                drop_columns.add(col2)

                print("  -->", col2, "can be dropped")


# Convert set to list
drop_columns = list(drop_columns)

print("\nColumns to be dropped:")
print(drop_columns)


# Create final dataset
final_df = df2f.drop(columns=drop_columns)

# Add target back
final_df['price'] = target2

print("\nFinal dataset:")
print(final_df.head())

print("\nRemaining columns:")
print(final_df.columns)


# Save final dataset
final_df.to_csv('housing_final.csv', index=False)
```
